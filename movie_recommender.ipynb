```python
movie=pd.read_csv(r"C:\\Users\\acrer\\Downloads\\movie\\tmdb_5000_movies.csv")
credit=pd.read_csv(r"C:\\Users\\acrer\\Downloads\\movie\\tmdb_5000_credits.csv")
```

```python
df=movie.merge(credit,on='title')
df.shape
```

```python
df.info()
```

```python
# fetching the columns we need
df=df[['movie_id','title','overview','genres','keywords','cast','crew']]
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
```

```python
#convert string representation of a list into list
import ast
def convert1(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
df['genres']=df['genres'].apply(convert1)
```

```python
df.iloc[0].genres
```

```python
#convert dictionary into a list
df['keywords']=df['keywords'].apply(convert1)
```

```python
#convert dictionary into list and get only top 3 actors name
def convert2(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter!=3:
          L.append(i['name'])
          counter+=1
        else:
           break  
    return L
df['cast']=df['cast'].apply(convert2)
```

```python
# to find director (where job=director)
def convert3(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
df['crew']=df['crew'].apply(convert3)
```

```python
df.head(3)
```

```python
#if overview contains missing value replace it with empty string
df['overview'] = df['overview'].fillna('')
print(df['overview'].apply(type).value_counts())
```

```python
#if value in overview is a string then split if not then use[]
df['overview'] = df['overview'].apply(
    lambda x: x.split() if isinstance(x, str) else []
)
```

```python
print(type(df['overview'].iloc[0]))
```

```python
#removing space between words
df['genres']=df['genres'].apply(lambda x:[i.replace(" ","")for i in x])
df['keywords']=df['keywords'].apply(lambda x:[i.replace(" ","")for i in x])
df['cast']=df['cast'].apply(lambda x:[i.replace(" ","")for i in x])
df['crew']=df['crew'].apply(lambda x:[i.replace(" ","")for i in x])
```

```python
cols = ['overview', 'genres', 'keywords', 'cast', 'crew']
for col in cols:
    print("\n", col)
    print(df[col].apply(type).value_counts())
```

```python
df['tags']=df['overview']+df['genres']+df['keywords']+df['cast']+df['crew']
new_df=df[['movie_id','title','genres','tags']]
new_df.head()
```

```python
#join every list in space
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
new_df.head()
```

```python
new_df['tags'][0]
```

```python
#fetch max times coming word in any movie in the column tag except stop words (is,are,the,in,and)
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=3000,stop_words='english')
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
print(cv.get_feature_names_out())
```

```python
#python library used for natural language processing,text analysis.Here we used this for stemming.
import nltk
from nltk.stem.porter import PorterStemmer # for standard english text preprocessing
ps=PorterStemmer()
```

```python
#ps.stem('Dancing') will give output Danc(root word)
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)    
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
#fetch max times coming word in any movie in the column tag except stop words (is,are,the,in,and)
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=3000,stop_words='english')
```

```python
#creating vectors from column tags and converting them to an array
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
print(cv.get_feature_names_out())
```

```python
#how closely two vectors align in direction(angle between two movies)
from sklearn.metrics.pairwise import cosine_similarity
cs=cosine_similarity(vectors)
cs.shape
```

```python
cs[0]
```

```python
# will give similarity with movie index in sorted order
sorted(list(enumerate(cs[0])),reverse = True,key=lambda x:x[1])[1:11]
```

```python
#function to recommend movie for given input (movie)
def recommend(movie):
    index = new_df[new_df['title']==movie].index[0]
    dist=cs[index]
    movie_list=sorted(list(enumerate(dist)),reverse = True,key=lambda x:x[1])[1:11]

    for i in movie_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Batman Begins')
```
